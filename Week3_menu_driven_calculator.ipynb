```python
while True:
    print("\n===== Menu Driven Calculator =====")
    print("Select the operation to be performed:")
    print("+. Addition")
    print("-. Subtraction")
    print("*. Multiplication")
    print("/. Division")
    print("0. Exit")

    choice = input("Enter the choice: ").strip()

    # Exit condition
    if choice == "0":
        print("Calculator closed.")
        break

    # Validate input choice
    if choice not in ["+", "-", "*", "/"]:
        print("Invalid choice. Please select a valid operation.")
        continue

    # Take two numbers as input
    try:
        num1 = float(input("Enter the first number: "))
        num2 = float(input("Enter the second number: "))
    except ValueError:
        print("Invalid number entered. Please try again.")
        continue

    # Perform the selected calculation
    if choice == "+":
        result = num1 + num2
    elif choice == "-":
        result = num1 - num2
    elif choice == "*":
        result = num1 * num2
    elif choice == "/":
        if num2 != 0:
            result = num1 / num2
        else:
            print("Error: Division by zero is not possible.")
            continue

    # Display the result
    print("Result:", result)
```
